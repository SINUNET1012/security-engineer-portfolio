```python
##Scapy란.
# 파이썬 기반 네트워크 패킷 제작. 조작, 전송, 분석하는 것이다.
# L2~L7 계층을 모두 지원하고, 사용자 입맛대로 패킷을 자유롭게 조작 / 구성이 가능

## Scapy를 쓰는 이유
# 다양한 기능을 하나의 도구로 수행 가능.
# scapy = 분석 + 제작 + 조작 + 스푸핑  모두 가능
# 공격자 입장에서 필요한 플래그 값 조합 / 헤더 변조 / 값 변조를 쉽게 가능하다.
# 잘 이용하면 대용량 패킷 생성과 분석에 유용하다.

## Scapy의 구조
# L2 : ether.ARP
# L3 : IP,IPv6
# L4 : TCP, UDP, ICMP
# L7 : DNS, HTTP, Raw 

# L1, L5, L6 는 왜 없나요?
# L1 물리계층은 scapy가 접근 불가. -> 하드웨어적이기 때문
# L5 세션층은 패킷 구조가 존재하지 않는다.
# L6 표현계층은 데이터 표현 방식 관련 계층이므로 관련 X
```
